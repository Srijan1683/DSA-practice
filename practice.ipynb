## I. Strings

### 1. Check if a string is palindrome:

```python
def isPalindrome(s):
    return s == s[::-1]

print(isPalindrome("abccba"))
print(isPalindrome("string"))


def isPalindrome2(x):
    left = 0
    right = len(x) - 1
    
    while left < right:
        if x[left] != x[right]:
            return False
        left += 1
        right -= 1
        
    return True

print(isPalindrome("string"))
print(isPalindrome("ssss"))
```

### 2. Reverse a string (without using any function):

```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("python"))


def reverse_string_2(d):
    result = ""
    for char in d:
        result = char + result
    return result

print(reverse_string_2("object"))
```

### 3. Count Vowels:

```python
def CountVowels(s):
    vowels = "aeiou"
    count = 0
    
    for char in s.lower():
        if char in vowels:
            count += 1
    return count

print(CountVowels("adamant"))
print(CountVowels("ADAMANT"))
```

### 4. Check Anagram:

```python
def anagram1(s1, s2):
    return sorted(s1) == sorted(s2)


def anagram2(s1, s2):
    if len(s1) != len(s2):
        return False
    
    count = {}
    
    for char in s1:
        count[char] = count.get(char, 0) + 1
        
    for char in s2:
        count[char] = count.get(char, 0) - 1
        
    for val in count.values():
        if val != 0:
            return False
        
    return True


print(anagram1("silent", "listen"))
print(anagram2("file", "life"))
```

### 5. Remove duplicates from a string:

```python
def removedupes(s):
    new = ""
    for char in s:
        if char not in new:
            new += char
        
    return new


def removedupesagain(s):
    seen = set()
    new = ""
    
    for char in s:
        if char not in seen:
            seen.add(char)
            new += char
    
    return new

print(removedupes("assassination"))
print(removedupesagain("mathematics"))
```

### 6. Most Frequent character in a string:

```python
def frequentstring(s):
    count = {}
    for char in s:
        count[char] = count.get(char, 0) + 1
        
    return max(count, key = count.get)

print(frequentstring("orthonormal"))
```

### 7. Check Pangram:

```python
def pangram(s):
    alphabets = "abcdefghijklmnopqrstuvwxyz"
    
    for char in alphabets:
        if char not in s:
            return False
    
    return True

print(pangram("string"))
print(pangram("The quick brown fox jumps over the lazy dog"))
```
