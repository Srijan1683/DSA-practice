## I. Strings

### 1. Check if a string is palindrome:

```python
def isPalindrome(s):
    return s == s[::-1]

print(isPalindrome("abccba"))
print(isPalindrome("string"))


def isPalindrome2(x):
    left = 0
    right = len(x) - 1
    
    while left < right:
        if x[left] != x[right]:
            return False
        left += 1
        right -= 1
        
    return True

print(isPalindrome("string"))
print(isPalindrome("ssss"))
```

### 2. Reverse a string (without using any function):

```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("python"))


def reverse_string_2(d):
    result = ""
    for char in d:
        result = char + result
    return result

print(reverse_string_2("object"))
```

### 3. Count Vowels:

```python
def CountVowels(s):
    vowels = "aeiou"
    count = 0
    
    for char in s.lower():
        if char in vowels:
            count += 1
    return count

print(CountVowels("adamant"))
print(CountVowels("ADAMANT"))
```

### 4. Check Anagram:

```python
def anagram1(s1, s2):
    return sorted(s1) == sorted(s2)


def anagram2(s1, s2):
    if len(s1) != len(s2):
        return False
    
    count = {}
    
    for char in s1:
        count[char] = count.get(char, 0) + 1
        
    for char in s2:
        count[char] = count.get(char, 0) - 1
        
    for val in count.values():
        if val != 0:
            return False
        
    return True


print(anagram1("silent", "listen"))
print(anagram2("file", "life"))
```

### 5. Remove duplicates from a string:

```python
def removedupes(s):
    new = ""
    for char in s:
        if char not in new:
            new += char
        
    return new


def removedupesagain(s):
    seen = set()
    new = ""
    
    for char in s:
        if char not in seen:
            seen.add(char)
            new += char
    
    return new

print(removedupes("assassination"))
print(removedupesagain("mathematics"))
```

### 6. Most Frequent character in a string:

```python
def frequentstring(s):
    count = {}
    for char in s:
        count[char] = count.get(char, 0) + 1
        
    return max(count, key = count.get)

print(frequentstring("orthonormal"))
```

### 7. Check Pangram:

```python
def pangram(s):
    alphabets = "abcdefghijklmnopqrstuvwxyz"
    
    for char in alphabets:
        if char not in s:
            return False
    
    return True

print(pangram("string"))
print(pangram("The quick brown fox jumps over the lazy dog"))
```

### 8. Replacing a word without using replace():

```python
def replaceword(s, old, new):
    words = []
    x = ""
    result = ""
    
    for char in s:
        if char == " ":
            words.append(x)
            x = ""
        else:
            x += char
            
    words.append(x)
    
    for i, word in enumerate(words):
        if word == old:
            result += new  
        else:
            result += word
            
        if i < len(words) - 1:
            result += " "
        
    return result
    
print(replaceword("hello world i am here", "am", "was"))
```

### 9. Convert Uppercase to Lowercase (and vice versa):

```python
def lowercase(s):
    new = ""
    
    for char in s:
        if 65 <= ord(char) <= 90:
            new += chr(ord(char) + 32)
        else:
            new += char
    
    return new

def uppercase(s):
    new2 = ""
    
    for char in s:
        if 97 <= ord(char) <= 122:
            new2 += chr(ord(char) - 32)
        else:
            new2 += char
    
    return new2
    
print(lowercase("GitHub"))
print(uppercase("arigato gozaimasu"))
```

### 10. Count Words in a String:

```python
def countwords(s):
    if not s.strip():
        return 0
    
    count = 0
    in_word = False
    
    for char in s:
        if char != " " and not in_word:
            count += 1
            in_word = True
        elif char == " ":
            in_word = False
    
    return count

print(countwords("Don't Go Insane"))
print(countwords(" hello world "))
```

### 11. Title Case:

```python
def titlecase(s):
    new = ""
    capitalise_next = True
    
    for char in s:
        if char == " ":
            capitalise_next = True
            new += char
            
        elif capitalise_next:
            new += chr(ord(char) - 32)
            capitalise_next = False
            
        else:
            new += char
            
    return new


print (titlecase("this is a python program"))
```

## II. Arrays/Lists

### 1. Remove Duplicates with and without using set():

```python
def removedupearray(nums):
    return list(set(nums))

print(removedupearray([1,2,3,4,5,5,5]))

def removedupearray2(nums):
    nums2 = []
    for num in nums:
        if num not in nums2:
            nums2.append(num)
            
    return nums2

print(removedupearray2([1,1,1,2,3,4,5,5,4]))
```

### 2. Find Second Largest Number:

```python
def findsecondlargest(nums):
    sortednums = sorted(nums)
    return sortednums[-2]

def findsecondlargest2(nums):
    first = float('-inf')
    second = float('-inf')
    
    for num in nums:
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
            
    return second

print(findsecondlargest([1,2,3,4,5]))
print(findsecondlargest2([2,4,1,3,0]))
```

### 3. Rotate a list by k positions:

```python
def rotatelist(nums, k):
    k = k % len(nums)
    return nums[-k:] + nums[:-k]

print(rotatelist([1,2,3,4,5], 2))
```

### 4. Two Sum:

```python
def twosum(nums, target):
    seen = {}
    
    for i, num in enumerate(nums):
        needed = target - num
        
        if needed in seen:
            return [seen[needed], i]
        
        seen[num] = i
        
print(twosum([1,2,3,4,5], 9))
```

### 5. Move Zeroes to the End:

```python
def movezeroes(nums):
    position = 0
    
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[position] = nums[i]
            position += 1
            
    while position < len(nums):
        nums[position] = 0
        position += 1
        
    return nums

print(movezeroes([0,1,4,0,5,0,1,0,3,0,0]))
```

### 6. Find Missing Number:

```python
def missingnumber(nums):
    n = len(nums)
    
    expected = (n + 1) * (n + 2) // 2
    actual = sum(nums)
    
    return expected - actual

print(missingnumber([1,2,3,5,6]))
```

### 7. Merge Two Sorted Arrays:

```python
def mergearray(nums1, nums2):
    i = 0
    j = 0
    result = []
    
    while i < len(nums1) and j < len(nums2):
        if nums1[i] < nums2[j]:
            result.append(nums1[i])
            i += 1
        
        else:
            result.append(nums2[j])
            j += 1
            
    while i < len(nums1):
        result.append(nums1[i])
        i += 1
        
    while j < len(nums2):
        result.append(nums2[j])
        j += 1
        
    return result

print(mergearray([1,2,3,4,5], [0,6,7,8]))
```

### 8. Flatten a Nested List:

```python
def flattenlist(nums):
    result = []
    
    for num in nums:
        if type(num) == list:
            for subnum in num:
                result.append(subnum)
                
        else:
            result.append(num)
            
    return result

print(flattenlist([[1, 2, 3, 4], 5, 6, 7, [8, 9]]))
```

## Linked Lists in Python

### Node Structure:

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

### Types of Linked Lists:

* Singly Linked Lists — each node points to the next only
* Doubly Linked Lists — each node has next and previous
* Circular Linked Lists — last node points back to the head

### Singly Linked List — Full Implementation

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None
        
        
    def append(self, data):
        new_node = Node(data)
        
        if not self.head:
            self.head = new_node
            return
        
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
        
    
    def prepend(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
```
